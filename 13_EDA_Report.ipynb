# 3. Statistical Findings

# Question 8 – What are the Central Tendency Findings?

## Step 1 – Understand the concept

Central tendency describes the typical or central value of a numerical feature.

The main measures are:

- Mean
- Median
- Mode

## Step 2 – Demonstrate the concept

For the Telco dataset, central tendency can be examined for:

- tenure
- MonthlyCharges
- TotalCharges

## Step 3 – Implementation the concept

Calculate mean and median for the important numerical features.

```python
numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

for column in numerical_features:
    data = pd.to_numeric(df[column], errors="coerce")

    print(f"\n--- {column} ---")
    print("Mean:", round(data.mean(), 2))
    print("Median:", round(data.median(), 2))
```

## Observation

Mean and median values help describe the typical customer behavior for each numerical feature.

## Insight

Large differences between mean and median may indicate skewed distributions or the presence of extreme observations.

# Question 9 – What are the Dispersion Findings?

## Step 1 – Understand the concept

Dispersion describes how spread out the values are.

Important measures include:

- Range
- Variance
- Standard deviation
- Interquartile range


## Step 2 – Demonstrate the concept

We can calculate these measures for the main numerical features.


## Step 3 – Implementation the concept

Calculate range, standard deviation, and IQR.

```python
for column in numerical_features:

    data = pd.to_numeric(df[column], errors="coerce").dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    print(f"\n--- {column} ---")
    print("Range:", round(data.max() - data.min(), 2))
    print("Standard Deviation:", round(data.std(), 2))
    print("IQR:", round(IQR, 2))
```

## Observation

The numerical features have different levels of spread.

## Insight

Understanding dispersion helps identify which variables have high variability and may require additional investigation during preprocessing and modeling.

# Question 10 – What are the Distribution Findings?

## Step 1 – Understand the concept

Distribution analysis describes the shape and behavior of numerical data.

Important aspects include:

- Symmetry
- Skewness
- Kurtosis
- Center
- Spread


## Step 2 – Demonstrate the concept

We can use skewness to understand whether numerical features are approximately symmetric or skewed.


## Step 3 – Implementation the concept

Calculate skewness for each numerical feature.

```python
for column in numerical_features:

    data = pd.to_numeric(df[column], errors="coerce")

    skewness = data.skew()

    if skewness > 0.5:
        distribution = "Right-Skewed"
    elif skewness < -0.5:
        distribution = "Left-Skewed"
    else:
        distribution = "Approximately Symmetric"

    print(
        f"{column}: "
        f"Skewness = {skewness:.2f} → {distribution}"
    )
```

## Observation

The numerical features may have different distribution shapes.

## Insight

Skewed features may require transformation during preprocessing depending on the requirements of the future machine learning model.

No transformation is performed in this notebook.

# Question 11 – What are the Outlier Findings?


## Step 1 – Understand the concept

An outlier is an observation that differs significantly from the general pattern of a feature.

Outliers may represent genuine customer behavior or data-quality problems.


## Step 2 – Demonstrate the concept

The IQR method can be used to identify potential outliers in numerical features.


## Step 3 – Implementation the concept

Identify potential outliers without removing them.

```python
for column in numerical_features:

    data = pd.to_numeric(df[column], errors="coerce").dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower_bound) |
        (data > upper_bound)
    ]

    print(f"\n--- {column} ---")
    print("Potential Outliers:", len(outliers))
```

## Observation

Potential outliers were identified using the IQR method.

## Insight

Outliers should not be automatically removed.

During preprocessing, each important outlier should be evaluated to determine whether it is a genuine observation, data-entry error, or an extreme value requiring transformation.
